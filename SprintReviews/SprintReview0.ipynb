```
using System;

// родительский класс Растение
class Plant
{
    public string Name;
    public string Type;
    public float Height;

    // метод цветения
    public virtual void Bloom()
    {
        Console.WriteLine("Растение " + Name + " цветет.");
    }

    // метод роста
    public void Grow(float growth)
    {
        Height = Height + growth;
        Console.WriteLine(Name + " выросло. Новая высота: " + Height + " см.");
    }
}

// класс Цветок наследуется от Растения
class Flower : Plant
{
    public string Color;
    public int Petals;

    // своя версия метода цветения
    public override void Bloom()
    {
        Console.WriteLine("Цветок " + Name + " распускается и радует глаз");
    }

    // метод увядания
    public void Wither()
    {
        Console.WriteLine("Цветок " + Name + " начинает увядать.");
    }
}

// класс Дерево наследуется от Растения
class Tree : Plant
{
    public int Age;
    public string LeafType;

    // своя версия метода цветения
    public override void Bloom()
    {
        Console.WriteLine("Дерево " + Name + " покрывается новыми листьями");
    }

    // метод увеличения возраста
    public void GrowTree(int years)
    {
        Age = Age + years;
        Console.WriteLine("Возраст дерева " + Name + ": " + Age + " лет.");
    }
}

// пример цветка
Flower flower = new Flower();

flower.Name = "Роза";
flower.Type = "Цветок";
flower.Height = 40;
flower.Color = "Красный";
flower.Petals = 20;

// пример дерева
Tree tree = new Tree();

tree.Name = "Береза";
tree.Type = "Дерево";
tree.Height = 300;
tree.Age = 5;
tree.LeafType = "Лиственные";

// проверка работы цветка
Console.WriteLine("ЦВЕТОК");

flower.Bloom();
flower.Grow(10);
flower.Wither();

Console.WriteLine();

// проверка работы дерева
Console.WriteLine("ДЕРЕВО");

tree.Bloom();
tree.Grow(50);
tree.GrowTree(2);

Console.WriteLine();

// полиморфизм
Console.WriteLine("ПОЛИМОРФИЗМ");

Plant plant1 = flower;
Plant plant2 = tree;

plant1.Bloom();
plant2.Bloom();
```
