## задание 1. класс Product

- усл: поля name, price, quantity сделать private, category - protected, description - internal. 
- методы:  GetTotalPrice (общая стоимость всех единиц товара), UpdateQuantity (обновить количество), GetProductInfo (строка с описанием товара). На все поля - геттеры и сеттеры.

```
using System;

public class Product
{
    // поля
    private string name;
    private decimal price;
    private int quantity;

    // protected 
    protected string category;

    // internal 
    internal string description;

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    //геттеры и сеттеры (через св-ва)
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set
        {
            if (value >= 0)
                price = value;
            else
                Console.WriteLine("Цена не может быть отрицательной.");
        }
    }

    // сеттер приват. кол-во меняется только через UpdateQuantity()
    public int Quantity
    {
        get { return quantity; }
        private set { quantity = value; }
    }

    protected string Category
    {
        get { return category; }
        set { category = value; }
    }

    internal string Description
    {
        get { return description; }
        set { description = value; }
    }

    // методы
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        int newQuantity = quantity + amount;
        if (newQuantity < 0)
        {
            Console.WriteLine($"Ошибка: недостаточно товара \"{name}\" на складе.");
            return;
        }
        quantity = newQuantity;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name} | Цена: {price:C} | Кол-во: {quantity} | Категория: {category} | Описание: {description}";
    }
}

Console.WriteLine("Класс Product объявлен.");
```

## задание 2. класс Seller

- усл: поля name, employeeId сделать private, salary - protected, contactInfo - internal. 
- методы - AddProduct (добавить товар в список продавца), SellProduct (уменьшить количество товара), GetSellerInfo (информация о продавце).

```
using System.Collections.Generic;

public class Seller
{
    private string name;
    private string employeeId;

    // protected
    protected decimal salary;

    // internal
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    protected decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    internal string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public List<Product> Products
    {
        get { return products; }
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар \"{product.Name}\" добавлен продавцу {name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (!products.Contains(product))
        {
            Console.WriteLine($"У продавца {name} нет такого товара в списке.");
            return;
        }
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продавец {name} продал {quantity} ед. товара \"{product.Name}\".");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name} | ID: {employeeId} | Контакты: {contactInfo}";
    }
}

Console.WriteLine("Класс Seller объявлен.");
```

## задание 3. класс Store

- усл: поля storeName, location сделать private, storeHours - public.
- методы: AddSeller (добавить продавца в магазин), ListProducts (перечислить все товары в магазине), GetStoreInfo (информация о магазине).

```
using System.Collections.Generic;

public class Store
{
    private string storeName;
    private string location;

    // публичное поле
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин \"{storeName}\".");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине \"{storeName}\":");
        foreach (var seller in sellers)
        {
            foreach (var product in seller.Products)
            {
                Console.WriteLine(" - " + product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName} | Адрес: {location} | Часы работы: {storeHours} | Продавцов: {sellers.Count}";
    }
}

Console.WriteLine("Класс Store объявлен.");
```
