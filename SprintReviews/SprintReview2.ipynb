```
using System;

// интерфейс для отчета
public interface IReportable
{
    string GenerateReport();
}

// интерфейс для рук-ля команды
public interface ITeamLeader
{
    string[] GetTeamMembers();
}

// базовый класс Employee (сотрудник)
public class Employee
{
    public string Name { get; set; }       // имя
    public string Position { get; set; }   // должность
    public double Salary { get; set; }     // зп

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    // вывод информации о сотруднике
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}, Должность: {Position}, Зарплата: {Salary}");
    }
}

//прост наследование: менеджер наследует от сотр-ка
public class Manager : Employee
{
    public string Department { get; set; }   // отдел

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    // переопр PrintInfo: добав отдел
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

// сложное наслед-е: ProjectManager наследуетт от manager (а manager от employee)
// множест-е насл: ProjectManager реализует два интерфейса
public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }   // название проекта
    public string[] Team { get; set; }        // члены команды

    public ProjectManager(string name, string position, double salary,
                          string department, string projectName, string[] team)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        Team = team;
    }

    // переопр PrintInfo: добавляем название проекта
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    // метод интерфейса IReportable
    public string GenerateReport()
    {
        return $"Отчет: {Name} отвечает за проект \"{ProjectName}\", в команде {Team.Length} чел.";
    }

    // метод интерфейса ITeamLeader
    public string[] GetTeamMembers()
    {
        return Team;
    }
}

// тестирование
// экземпляры каждого класса
Employee employee = new Employee("Иван Петров", "Бухгалтер", 50000);
Manager manager = new Manager("Анна Смирнова", "Менеджер", 80000, "Продажи");
string[] team = { "Мария", "Дмитрий", "Екатерина" };
ProjectManager projectManager = new ProjectManager("Олег Ковалев", "Проектный менеджер",
                                                   100000, "IT", "Мобильное приложение", team);

// вызов PrintInfo() для каждого экземпляра
Console.WriteLine("--- Employee ---");
employee.PrintInfo();

Console.WriteLine("\n--- Manager ---");
manager.PrintInfo();

Console.WriteLine("\n--- ProjectManager ---");
projectManager.PrintInfo();

// тест методов интерфейсов
Console.WriteLine("\n--- Методы интерфейсов ---");
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("Члены команды:");
foreach (string member in projectManager.GetTeamMembers())
{
    Console.WriteLine("- " + member);
}
```
