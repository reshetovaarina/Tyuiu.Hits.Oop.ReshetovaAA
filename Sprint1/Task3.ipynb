### Вариант задания № 16 - Система способов оплаты (PaymentMethod)


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс PaymentMethod, который представляет общий способ оплаты.
На основе этого класса разработать 3 производных класса: OnlinePayment (онлайн-оплата),
BankTransfer (банковский перевод) и CashPayment (оплата наличными), демонстрирующих
принципы наследования и полиморфизма.

Базовый класс PaymentMethod содержит:
- Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты (MethodName),
  Минимальная сумма (MinAmount)
- Методы: ProcessPayment(), CheckMinimumAmount(), GetPaymentDetails()

Производные классы:
1. OnlinePayment - добавляет URL платежной системы, переопределяет ProcessPayment()
2. BankTransfer - добавляет банковские данные, переопределяет CheckMinimumAmount()
3. CashPayment - добавляет место выдачи наличных, переопределяет GetPaymentDetails()

Дополнительное задание:
- Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

**Что сделано в дополнительном задании:
- Все атрибуты классов оформлены как свойства с геттерами (get) и сеттерами (set).
- У каждого класса есть конструктор, который записывает значения в свойства.
- Добавлен класс Customer (клиент). Его метод Pay() принимает объект PaymentMethod и вызывает его методы CheckMinimumAmount() и ProcessPayment() - так объекты взаимодействуют между собой.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class PaymentMethod
{
    // Приватные поля
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;

    // св-ва с геттерами и сеттерами
    public int PaymentMethodId
    {
        get { return _paymentMethodId; }
        set { _paymentMethodId = value; }
    }

    public string MethodName
    {
        get { return _methodName; }
        set { _methodName = value; }
    }

    public decimal MinAmount
    {
        get { return _minAmount; }
        set { _minAmount = value; }
    }

    // конструктор
    public PaymentMethod(int id, string name, decimal minAmount)
    {
        PaymentMethodId = id;
        MethodName = name;
        MinAmount = minAmount;
    }

    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Обработка платежа: {amount} через {MethodName}");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual string GetPaymentDetails()
    {
        return $"Способ оплаты: {MethodName}, ID: {PaymentMethodId}, Минимальная сумма: {MinAmount}";
    }
}

public class OnlinePayment : PaymentMethod
{
    private string _paymentUrl;

    public string PaymentUrl
    {
        get { return _paymentUrl; }
        set { _paymentUrl = value; }
    }

    public OnlinePayment(int id, string name, decimal minAmount, string paymentUrl)
        : base(id, name, minAmount)
    {
        PaymentUrl = paymentUrl;
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Оплата {amount} через онлайн-систему по адресу {PaymentUrl}");
    }
}

public class BankTransfer : PaymentMethod
{
    private string _bankData;

    public string BankData
    {
        get { return _bankData; }
        set { _bankData = value; }
    }

    public BankTransfer(int id, string name, decimal minAmount, string bankData)
        : base(id, name, minAmount)
    {
        BankData = bankData;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * 0.01m; // усл банковская комиссия 1%
        Console.WriteLine($"Проверка минимальной суммы с учётом комиссии: {commission}");
        return (amount - commission) >= MinAmount;
    }
}

public class CashPayment : PaymentMethod
{
    private string _cashPickupPoint;

    public string CashPickupPoint
    {
        get { return _cashPickupPoint; }
        set { _cashPickupPoint = value; }
    }

    public CashPayment(int id, string name, decimal minAmount, string cashPickupPoint)
        : base(id, name, minAmount)
    {
        CashPickupPoint = cashPickupPoint;
    }

    public override string GetPaymentDetails()
    {
        return base.GetPaymentDetails() + $", Место выдачи наличных: {CashPickupPoint}";
    }
}

// класс для взаим-я объектов
public class Customer
{
    public string Name { get; set; }
    public decimal Balance { get; set; }

    public Customer(string name, decimal balance)
    {
        Name = name;
        Balance = balance;
    }

    // клиент оплачивает покупку выбранным способом оплаты
    public void Pay(PaymentMethod method, decimal amount)
    {
        if (!method.CheckMinimumAmount(amount))
        {
            Console.WriteLine($"{Name}: платёж отклонён, сумма меньше минимальной.");
        }
        else if (amount > Balance)
        {
            Console.WriteLine($"{Name}: платёж отклонён, недостаточно средств.");
        }
        else
        {
            method.ProcessPayment(amount);
            Balance = Balance - amount;
            Console.WriteLine($"{Name}: платёж выполнен. Остаток на балансе: {Balance}");
        }
    }
}

// основная программа - создание объектов и демонстрация работы

PaymentMethod online = new OnlinePayment(1, "Онлайн-оплата", 100, "https://pay.example.com");
online.ProcessPayment(500);
Console.WriteLine(online.GetPaymentDetails());

Console.WriteLine();

PaymentMethod bank = new BankTransfer(2, "Банковский перевод", 200, "40817810000000000000");
bank.CheckMinimumAmount(1000);
Console.WriteLine(bank.GetPaymentDetails());

Console.WriteLine();

PaymentMethod cash = new CashPayment(3, "Наличные", 0, "ул. Мельникайте, д. 70");
Console.WriteLine(cash.GetPaymentDetails());

// взаимодействие объектов: клиент оплачивает разными способами
Console.WriteLine();
Console.WriteLine("Клиент оплачивает покупки");

Customer customer = new Customer("Анна", 2000);

customer.Pay(online, 500);   // онлайн-оплата
customer.Pay(bank, 150);     // банк: с учётом комиссии меньше 200 - отказ
customer.Pay(bank, 1000);    // банковский перевод
customer.Pay(cash, 300);     // наличные
customer.Pay(online, 5000);  // не хватает средств
```
