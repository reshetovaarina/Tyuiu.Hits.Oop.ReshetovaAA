### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

````
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с валидацией в сеттере
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException(nameof(Speed), "Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}

public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}

// Проверка
var vehicle = new Vehicle { Color = "Red" };
vehicle.Accelerate(60);
Console.WriteLine($"Vehicle: цвет = {vehicle.Color}, скорость = {vehicle.Speed}");

var car = new Car { Color = "Blue" };
car.Refuel(50);
car.Accelerate(90);
Console.WriteLine($"Car: цвет = {car.Color}, скорость = {car.Speed}, топливо = {car.FuelLevel}");

Vehicle.DisplayTotalVehicles();

// Проверка валидации в сеттере
try
{
    car.Speed = -5;
}
catch (ArgumentOutOfRangeException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса

````
public class Animal
{
    // Обычные атрибуты (у каждого объекта свои значения)
    private string _name;
    private int _age;
    private int _energy;

    // Статические атрибуты (общие для всего класса)
    public const int MaxEnergy = 100;
    public static int TotalAnimals { get; private set; } = 0;

    // Свойство Name с валидацией
    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    // Свойство Age с валидацией
    public int Age
    {
        get { return _age; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException(nameof(Age), "Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    // Энергия: читать можно всем, менять - только классу и наследникам,
    // значение всегда ограничено диапазоном 0..MaxEnergy
    public int Energy
    {
        get { return _energy; }
        protected set { _energy = Math.Clamp(value, 0, MaxEnergy); }
    }

    // Свойство только для чтения (вычисляется в геттере)
    public bool IsHungry
    {
        get { return Energy < 30; }
    }

    // Вид животного: читать можно всем, менять - только наследникам
    public string Species { get; protected set; } = "Животное";

    // Конструктор
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
        Energy = 50;
        TotalAnimals++;
    }

    // Обычные методы
    public void Eat(int food)
    {
        if (food <= 0)
            throw new ArgumentOutOfRangeException(nameof(food), "Количество еды должно быть положительным!");
        Energy += food;
    }

    public virtual string MakeSound()
    {
        return "...";
    }

    public virtual void Move()
    {
        Energy -= 10;
    }

    public string GetInfo()
    {
        return $"{Species} {Name}, возраст: {Age}, энергия: {Energy}, голоден: {(IsHungry ? "да" : "нет")}";
    }

    // Статический метод
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    private string _breed;

    public string Breed
    {
        get { return _breed; }
        set { _breed = string.IsNullOrWhiteSpace(value) ? "Дворняжка" : value; }
    }

    // Статический атрибут производного класса
    public static int TotalDogs { get; private set; } = 0;

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Species = "Собака";
        Breed = breed;
        TotalDogs++;
    }

    public override string MakeSound()
    {
        return "Гав!";
    }

    // Собственный метод
    public void Fetch()
    {
        Energy -= 20;
        Console.WriteLine($"{Name} приносит палку. Энергия: {Energy}");
    }

    // Статический метод
    public static void DisplayTotalDogs()
    {
        Console.WriteLine($"Собак: {TotalDogs}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public static int TotalCats { get; private set; } = 0;

    public Cat(string name, int age, bool isIndoor) : base(name, age)
    {
        Species = "Кошка";
        IsIndoor = isIndoor;
        TotalCats++;
    }

    public override string MakeSound()
    {
        return "Мяу!";
    }

    // Собственный метод
    public void Purr()
    {
        Energy += 5;
        Console.WriteLine($"{Name} мурлычет. Энергия: {Energy}");
    }

    public static void DisplayTotalCats()
    {
        Console.WriteLine($"Кошек: {TotalCats}");
    }
}

public class Bird : Animal
{
    private double _wingSpan;

    public double WingSpan
    {
        get { return _wingSpan; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException(nameof(WingSpan), "Размах крыльев должен быть положительным!");
            _wingSpan = value;
        }
    }

    public static int TotalBirds { get; private set; } = 0;

    public Bird(string name, int age, double wingSpan) : base(name, age)
    {
        Species = "Птица";
        WingSpan = wingSpan;
        TotalBirds++;
    }

    public override string MakeSound()
    {
        return "Чирик!";
    }

    // Переопределение: птица тратит на перемещение больше энергии
    public override void Move()
    {
        Energy -= 15;
        Console.WriteLine($"{Name} летит. Энергия: {Energy}");
    }

    public static void DisplayTotalBirds()
    {
        Console.WriteLine($"Птиц: {TotalBirds}");
    }
}

// Проверка
var dog = new Dog("Рекс", 3, "Овчарка");
var cat = new Cat("Мурка", 2, true);
var bird = new Bird("Кеша", 1, 0.3);

Animal[] animals = { dog, cat, bird };
foreach (var a in animals)
{
    Console.WriteLine($"{a.GetInfo()} | звук: {a.MakeSound()}");
}

dog.Fetch();
cat.Purr();
bird.Move();
dog.Eat(25);

Console.WriteLine();
Console.WriteLine(dog.GetInfo());
Console.WriteLine(cat.GetInfo());
Console.WriteLine(bird.GetInfo());

Console.WriteLine();
Animal.DisplayTotalAnimals();
Dog.DisplayTotalDogs();
Cat.DisplayTotalCats();
Bird.DisplayTotalBirds();

// Проверка валидации в сеттере
try
{
    dog.Age = -1;
}
catch (ArgumentOutOfRangeException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}
````
